1. Рассчитать средний возраст n человек, n вводится с белками

```python
n = int(input("Введите количество человек (n): "))

print("Введите возрасты через пробел:")
ages_str = input().split()

ages = []
for age in ages_str:
    ages.append(int(age))

average_age = sum(ages) / n

print(f"Средний возраст: {average_age:.2f}")
```

2. Во входной строке кода цвета (например, «a0351f») определите цвет, ближайший к красному, синему или зеленому. Вывести, что цвет серый, если все равны, но цвет не черный и не белый. Вывести «черный» или «белый», если код 000000 или ffffff соответственно.

```python
color_code = input("Введите HEX-код цвета (например, a0351f): ").lower()

if color_code == "000000":
    print("черный")
elif color_code == "ffffff":
    print("белый")
else:
    r = int(color_code[0:2], 16)
    g = int(color_code[2:4], 16)
    b = int(color_code[4:6], 16)

    if r == g and g == b:
        print("серый")
    else:
        if r > g and r > b:
            print("красный")
        elif g > r and g > b:
            print("зеленый")
        else:
            print("синий")
```
